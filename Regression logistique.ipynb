```python
#importer les librairies 
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression 
```

```python
#importer le dataset 
dataset = pd.read_csv('diabetes-dataset.csv')
```

```python
#visualisation des donnees
dataset.head()
```

```python
#Afficher le dataset (Glucose - Outcome)
plt.scatter(dataset.Glucose , dataset.Outcome)
```

```python
# Definir notre variable dependante y et notre variable independante X
X = dataset.iloc[:, :-1].values
y = dataset.iloc[:, -1].values
```

```python
# Diviser le dataset en training set et test set
X_train ,X_test, y_train , y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

```python
# Construction du modele 
classifier = LogisticRegression ( solver = 'liblinear' , random_state = 0 ) 
classifier.fit(X_train, y_train)
```

```python
# Faire de nouvelle predictions 
y_pred = classifier.predict(X_test)
classifier.score(X_test , y_test)
```
